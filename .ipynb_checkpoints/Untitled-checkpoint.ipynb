```python
import os
import time
import pickle
import warnings
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, label_binarize
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    precision_recall_curve
)
from sklearn.neighbors import NearestNeighbors

from imblearn.over_sampling import RandomOverSampler, SMOTE

warnings.filterwarnings("ignore")
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

print("Libraries imported successfully.")


DATA_PATH = "dry+bean+dataset/DryBeanDataset/Dry_Bean_Dataset.xlsx"

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"Could not find {DATA_PATH}. Put the Excel file beside this notebook "
        "or update DATA_PATH with the correct file location."
    )

df = pd.read_excel(DATA_PATH)
df.columns = df.columns.astype(str).str.strip()

# Find the target column without depending on capitalisation.
target_matches = [col for col in df.columns if col.lower() == "class"]
if not target_matches:
    raise ValueError("A target column named 'Class' was not found. Check the Excel column names.")

target_col = target_matches[0]

print("Dataset shape:", df.shape)
print("\nFirst five rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values per column:")
display(df.isnull().sum().to_frame("Missing values"))
```

```python
# Remove exact duplicate rows, if any.
duplicate_count = df.duplicated().sum()
print("Exact duplicate rows found:", duplicate_count)

df = df.drop_duplicates().reset_index(drop=True)

X = df.drop(columns=[target_col]).copy()
y = df[target_col].astype(str).copy()

# Ensure feature columns are numeric.
for col in X.columns:
    X[col] = pd.to_numeric(X[col], errors="coerce")

print("Dataset shape after duplicate removal:", df.shape)
print("\nNumber of classes:", y.nunique())
print("\nClass distribution:")
class_counts = y.value_counts().sort_values(ascending=False)
display(class_counts.to_frame("Number of samples"))

imbalance_ratio = class_counts.max() / class_counts.min()
print(f"Majority-to-minority ratio: {imbalance_ratio:.2f}:1")

plt.figure(figsize=(10, 5))
sns.barplot(x=class_counts.index, y=class_counts.values)
plt.title("Class Distribution in the Original Dataset")
plt.xlabel("Bean class")
plt.ylabel("Number of samples")
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

```python
# First separate 70% training and 30% temporary data.
X_train_df, X_temp_df, y_train, y_temp = train_test_split(
    X, y,
    test_size=0.30,
    random_state=RANDOM_STATE,
    stratify=y
)

# Split the temporary data equally into validation and test (15% each overall).
X_val_df, X_test_df, y_val, y_test = train_test_split(
    X_temp_df, y_temp,
    test_size=0.50,
    random_state=RANDOM_STATE,
    stratify=y_temp
)

print("Training rows:", len(X_train_df))
print("Validation rows:", len(X_val_df))
print("Test rows:", len(X_test_df))

# Confirm that the original row indices do not overlap.
assert set(X_train_df.index).isdisjoint(set(X_val_df.index))
assert set(X_train_df.index).isdisjoint(set(X_test_df.index))
assert set(X_val_df.index).isdisjoint(set(X_test_df.index))
print("Leakage check: no original row index is shared between splits.")

print("\nTraining class counts:")
display(y_train.value_counts().sort_index().to_frame("Training samples"))
```

```python
imputer = SimpleImputer(strategy="median")
scaler = StandardScaler()

X_train_imputed = imputer.fit_transform(X_train_df)
X_val_imputed = imputer.transform(X_val_df)
X_test_imputed = imputer.transform(X_test_df)

X_train_scaled = scaler.fit_transform(X_train_imputed)
X_val_scaled = scaler.transform(X_val_imputed)
X_test_scaled = scaler.transform(X_test_imputed)

feature_names = X.columns.tolist()
class_names = sorted(y.unique())

print("Training feature matrix:", X_train_scaled.shape)
print("Validation feature matrix:", X_val_scaled.shape)
print("Test feature matrix:", X_test_scaled.shape)
print("Classes:", class_names)
```

```python
# A common classifier configuration for a fair comparison.
def make_random_forest(class_weight=None):
    return RandomForestClassifier(
        n_estimators=300,
        random_state=RANDOM_STATE,
        n_jobs=-1,
        class_weight=class_weight
    )

# Baseline: no resampling and no class weighting.
baseline_model = make_random_forest()
baseline_model.fit(X_train_scaled, y_train)

# Random oversampling: duplicate minority-class training rows.
ros = RandomOverSampler(random_state=RANDOM_STATE)
X_ros, y_ros = ros.fit_resample(X_train_scaled, y_train)

ros_model = make_random_forest()
ros_model.fit(X_ros, y_ros)

# SMOTE: create synthetic minority-class training rows.
# Set k_neighbors safely based on the smallest class count in the training set.
smallest_train_class = y_train.value_counts().min()
if smallest_train_class < 2:
    raise ValueError("SMOTE needs at least 2 training examples in every class.")
smote_k = min(5, smallest_train_class - 1)

smote = SMOTE(random_state=RANDOM_STATE, k_neighbors=smote_k)
X_smote, y_smote = smote.fit_resample(X_train_scaled, y_train)

smote_model = make_random_forest()
smote_model.fit(X_smote, y_smote)

# Class weighting: no resampling; increase the penalty for mistakes on rare classes.
weighted_model = make_random_forest(class_weight="balanced")
weighted_model.fit(X_train_scaled, y_train)

models = {
    "Baseline": baseline_model,
    "Random Oversampling": ros_model,
    "SMOTE": smote_model,
    "Class Weighting": weighted_model
}

print("All four models have been trained.")
print("\nTraining class counts after each resampling method:")
print("Original:", Counter(y_train))
print("Random oversampling:", Counter(y_ros))
print("SMOTE:", Counter(y_smote))
```

```python
# SMOTE keeps the original training rows first and appends generated rows.
n_original_train = len(X_train_scaled)
X_synthetic_scaled = np.asarray(X_smote)[n_original_train:]
y_synthetic = np.asarray(y_smote)[n_original_train:]

print("Number of synthetic rows generated:", len(X_synthetic_scaled))
print("Synthetic class counts:", Counter(y_synthetic))

# Convert synthetic data back to the original feature scale for inspection.
X_synthetic_original_scale = scaler.inverse_transform(X_synthetic_scaled)
synthetic_df = pd.DataFrame(X_synthetic_original_scale, columns=feature_names)
synthetic_df["Class"] = y_synthetic

print("\nFirst five generated examples:")
display(synthetic_df.head())

# Check feature ranges against the original training data.
train_min = X_train_df.min()
train_max = X_train_df.max()
outside_range = (
    (synthetic_df[feature_names] < train_min) |
    (synthetic_df[feature_names] > train_max)
).sum()

range_check = pd.DataFrame({
    "Synthetic values outside training range": outside_range,
    "Total synthetic rows": len(synthetic_df)
})
display(range_check)

# Nearest-neighbour distance to real training examples of the rarest class.
minority_class = y_train.value_counts().idxmin()
real_minority_scaled = X_train_scaled[np.asarray(y_train == minority_class)]

synthetic_minority_scaled = X_synthetic_scaled[
    np.asarray(y_synthetic == minority_class)
]

if len(synthetic_minority_scaled) > 0:
    nn = NearestNeighbors(n_neighbors=1)
    nn.fit(real_minority_scaled)
    distances, _ = nn.kneighbors(synthetic_minority_scaled)

    print(f"Rarest training class: {minority_class}")
    print("Median nearest-neighbour distance:", round(float(np.median(distances)), 4))
    print("95th percentile nearest-neighbour distance:", round(float(np.percentile(distances, 95)), 4))
else:
    print("No synthetic rows were generated for the rarest class.")
```

```python
minority_class = y_train.value_counts().idxmin()
print("Minority class used for minority precision/recall:", minority_class)

def calculate_metrics(model, X_eval, y_eval):
    y_pred = model.predict(X_eval)
    y_prob = model.predict_proba(X_eval)

    # Macro PR-AUC: average precision for each class, treating it as one-vs-rest.
    y_eval_binary = label_binarize(y_eval, classes=class_names)
    macro_pr_auc = average_precision_score(
        y_eval_binary, y_prob, average="macro"
    )

    # Per-class false-positive rate, then macro-average.
    cm = confusion_matrix(y_eval, y_pred, labels=class_names)
    fprs = []
    for i in range(len(class_names)):
        tp = cm[i, i]
        fn = cm[i, :].sum() - tp
        fp = cm[:, i].sum() - tp
        tn = cm.sum() - (tp + fn + fp)
        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0.0
        fprs.append(fpr)

    return {
        "Accuracy": accuracy_score(y_eval, y_pred),
        "Macro Precision": precision_score(y_eval, y_pred, average="macro", zero_division=0),
        "Macro Recall": recall_score(y_eval, y_pred, average="macro", zero_division=0),
        "Macro-F1": f1_score(y_eval, y_pred, average="macro", zero_division=0),
        "Minority Precision": precision_score(
            y_eval, y_pred, labels=[minority_class], average="macro", zero_division=0
        ),
        "Minority Recall": recall_score(
            y_eval, y_pred, labels=[minority_class], average="macro", zero_division=0
        ),
        "Macro PR-AUC": macro_pr_auc,
        "Macro FPR": float(np.mean(fprs))
    }
```

```python
validation_results = []

for name, model in models.items():
    row = {"Experiment": name}
    row.update(calculate_metrics(model, X_val_scaled, y_val))
    validation_results.append(row)

validation_df = pd.DataFrame(validation_results).set_index("Experiment")
display(validation_df.round(4))
```

```python
test_results = []

for name, model in models.items():
    row = {"Experiment": name}
    row.update(calculate_metrics(model, X_test_scaled, y_test))
    test_results.append(row)

test_results_df = pd.DataFrame(test_results).set_index("Experiment")
display(test_results_df.round(4))

# Save the final metrics for use in the report.
test_results_df.to_csv("test_metrics_comparison.csv")
print("Saved: test_metrics_comparison.csv")
```

```python
metrics_to_plot = [
    "Macro-F1", "Minority Precision", "Minority Recall", "Macro PR-AUC"
]

ax = test_results_df[metrics_to_plot].plot(
    kind="bar", figsize=(12, 6), ylim=(0, 1), rot=15
)
ax.set_title("Test-set comparison of imbalance-handling strategies")
ax.set_ylabel("Score")
ax.set_xlabel("Training strategy")
ax.legend(loc="lower right")
plt.tight_layout()
plt.savefig("metric_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 13))
axes = axes.ravel()

for ax, (name, model) in zip(axes, models.items()):
    y_pred = model.predict(X_test_scaled)
    cm = confusion_matrix(y_test, y_pred, labels=class_names)

    sns.heatmap(
        cm, annot=True, fmt="d", cmap="Blues",
        xticklabels=class_names, yticklabels=class_names, ax=ax
    )
    ax.set_title(name)
    ax.set_xlabel("Predicted class")
    ax.set_ylabel("Actual class")
    ax.tick_params(axis="x", rotation=40)
    ax.tick_params(axis="y", rotation=0)

plt.tight_layout()
plt.savefig("confusion_matrices.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
for name, model in models.items():
    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)
    y_pred = model.predict(X_test_scaled)
    print(classification_report(
        y_test, y_pred,
        labels=class_names,
        zero_division=0
    ))
    
```

```python
plt.figure(figsize=(8, 6))

minority_column = class_names.index(minority_class)
y_test_minority_binary = (np.asarray(y_test) == minority_class).astype(int)

for name, model in models.items():
    probabilities = model.predict_proba(X_test_scaled)[:, minority_column]
    precision, recall, _ = precision_recall_curve(
        y_test_minority_binary, probabilities
    )
    plt.plot(recall, precision, label=name)

plt.title(f"Precision-Recall Curves: Minority Class ({minority_class})")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("minority_precision_recall.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Inference timing uses the same fixed test sample for every model.
sample_size = min(500, len(X_test_scaled))
X_timing = X_test_scaled[:sample_size]

computational_rows = []

for name, model in models.items():
    # Warm-up call.
    model.predict_proba(X_timing[:min(20, sample_size)])

    repeats = 20
    start = time.perf_counter()
    for _ in range(repeats):
        model.predict_proba(X_timing)
    elapsed = time.perf_counter() - start

    latency_ms_per_image = (elapsed / repeats / sample_size) * 1000
    model_size_mb = len(pickle.dumps(model)) / (1024 ** 2)

    computational_rows.append({
        "Experiment": name,
        "Inference latency (ms/image)": latency_ms_per_image,
        "Model size (MB)": model_size_mb
    })

computational_df = pd.DataFrame(computational_rows).set_index("Experiment")
display(computational_df.round(4))
computational_df.to_csv("computational_cost_comparison.csv")
print("Saved: computational_cost_comparison.csv")
```

```python
prediction_examples = X_test_df.copy().reset_index(drop=True)
prediction_examples["Actual class"] = np.asarray(y_test)

for name, model in models.items():
    prediction_examples[name + " prediction"] = model.predict(X_test_scaled)

# Show cases where at least one strategy makes an incorrect prediction.
prediction_columns = [name + " prediction" for name in models.keys()]
incorrect_mask = np.zeros(len(prediction_examples), dtype=bool)

for col in prediction_columns:
    incorrect_mask |= prediction_examples[col].to_numpy() != prediction_examples["Actual class"].to_numpy()

error_examples = prediction_examples.loc[incorrect_mask].head(10)
display(error_examples)

error_examples.to_csv("representative_test_errors.csv", index=False)
print("Saved: representative_test_errors.csv")
```
