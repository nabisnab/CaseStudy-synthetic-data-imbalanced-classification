```python

# =====================================================================
# Synthetic Data for Imbalanced Classification (UCI Dry Bean Dataset)
# Baseline vs Random Oversampling vs SMOTE vs Class Weighting
# =====================================================================
import os, sys, time, pickle, warnings
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn, imblearn

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, label_binarize
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    average_precision_score, confusion_matrix, classification_report,
    precision_recall_curve
)
from sklearn.neighbors import NearestNeighbors
from imblearn.over_sampling import RandomOverSampler, SMOTE

warnings.filterwarnings("ignore")
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

# %% [markdown]
# ## Output 1 — Dataset Overview

# %%
# ---------------------------------------------------------------------
# 1. Load and inspect the dataset
# ---------------------------------------------------------------------
DATA_PATH = "dry+bean+dataset/DryBeanDataset/Dry_Bean_Dataset.xlsx"
if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"Could not find {DATA_PATH}. Put the Excel file beside this notebook "
        "or update DATA_PATH with the correct file location."
    )

df = pd.read_excel(DATA_PATH)
df.columns = df.columns.astype(str).str.strip()

target_matches = [c for c in df.columns if c.lower() == "class"]
if not target_matches:
    raise ValueError("A target column named 'Class' was not found.")
target_col = target_matches[0]

print("=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)
print(f"Number of rows:    {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
print(f"Class labels:      {sorted(df[target_col].unique())}")
print()
display(df.head())
print("\nMissing values per column:")
display(df.isnull().sum().to_frame("Missing values"))

# %% [markdown]
# ## Output 2 — Original Class Distribution Graph

# %%
# ---------------------------------------------------------------------
# 2. Clean data and inspect class imbalance
# ---------------------------------------------------------------------
print("Exact duplicate rows found:", df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)

X = df.drop(columns=[target_col]).copy()
y = df[target_col].astype(str).copy()
for col in X.columns:
    X[col] = pd.to_numeric(X[col], errors="coerce")

print("Dataset shape after duplicate removal:", df.shape)
print("Number of classes:", y.nunique())
class_counts = y.value_counts().sort_values(ascending=False)
display(class_counts.to_frame("Number of samples"))
print(f"Majority-to-minority ratio: {class_counts.max() / class_counts.min():.2f}:1")

plt.figure(figsize=(10, 5))
sns.barplot(x=class_counts.index, y=class_counts.values)
plt.title("Output 2: Class Distribution in the Original Dataset")
plt.xlabel("Bean class"); plt.ylabel("Number of samples")
plt.xticks(rotation=35); plt.tight_layout(); plt.show()

# %%
# ---------------------------------------------------------------------
# 3. Stratified split BEFORE resampling (70 / 15 / 15)
# ---------------------------------------------------------------------
X_train_df, X_temp_df, y_train, y_temp = train_test_split(
    X, y, test_size=0.30, random_state=RANDOM_STATE, stratify=y
)
X_val_df, X_test_df, y_val, y_test = train_test_split(
    X_temp_df, y_temp, test_size=0.50, random_state=RANDOM_STATE, stratify=y_temp
)
print("Training rows:", len(X_train_df))
print("Validation rows:", len(X_val_df))
print("Test rows:", len(X_test_df))

assert set(X_train_df.index).isdisjoint(X_val_df.index)
assert set(X_train_df.index).isdisjoint(X_test_df.index)
assert set(X_val_df.index).isdisjoint(X_test_df.index)
print("Leakage check: no original row index is shared between splits.")
display(y_train.value_counts().sort_index().to_frame("Training samples"))

# %%
# ---------------------------------------------------------------------
# 4. Preprocess (fit on training data only)
# ---------------------------------------------------------------------
imputer = SimpleImputer(strategy="median")
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(imputer.fit_transform(X_train_df))
X_val_scaled = scaler.transform(imputer.transform(X_val_df))
X_test_scaled = scaler.transform(imputer.transform(X_test_df))

feature_names = X.columns.tolist()
class_names = sorted(y.unique())
print("Classes:", class_names)

# %%
# ---------------------------------------------------------------------
# 5. Train the four strategies
# ---------------------------------------------------------------------
def make_random_forest(class_weight=None):
    return RandomForestClassifier(
        n_estimators=300, random_state=RANDOM_STATE,
        n_jobs=-1, class_weight=class_weight
    )

# Baseline
baseline_model = make_random_forest().fit(X_train_scaled, y_train)

# Random oversampling
X_ros, y_ros = RandomOverSampler(random_state=RANDOM_STATE).fit_resample(X_train_scaled, y_train)
ros_model = make_random_forest().fit(X_ros, y_ros)

# SMOTE
smallest_train_class = y_train.value_counts().min()
if smallest_train_class < 2:
    raise ValueError("SMOTE needs at least 2 training examples in every class.")
smote_k = min(5, smallest_train_class - 1)
X_smote, y_smote = SMOTE(random_state=RANDOM_STATE, k_neighbors=smote_k).fit_resample(X_train_scaled, y_train)
smote_model = make_random_forest().fit(X_smote, y_smote)

# Class weighting
weighted_model = make_random_forest(class_weight="balanced").fit(X_train_scaled, y_train)

models = {
    "Baseline": baseline_model,
    "Random Oversampling": ros_model,
    "SMOTE": smote_model,
    "Class Weighting": weighted_model,
}
print("All four models have been trained.")
print("Original:", Counter(y_train))
print("Random oversampling:", Counter(y_ros))
print("SMOTE:", Counter(y_smote))

# %%
# ---------------------------------------------------------------------
# 6. Check the synthetic samples created by SMOTE
# ---------------------------------------------------------------------
# NOTE: imbalanced-learn appends synthetic samples after the original
# samples.  This is the documented behaviour of SMOTE.fit_resample().
n_original_train = len(X_train_scaled)
X_synthetic_scaled = np.asarray(X_smote)[n_original_train:]
y_synthetic = np.asarray(y_smote)[n_original_train:]

print("\nNumber of synthetic rows generated:", len(X_synthetic_scaled))
print("Synthetic class counts:", Counter(y_synthetic))

synthetic_df = pd.DataFrame(scaler.inverse_transform(X_synthetic_scaled), columns=feature_names)
synthetic_df["Class"] = y_synthetic
display(synthetic_df.head())

train_min, train_max = X_train_df.min(), X_train_df.max()
outside_range = ((synthetic_df[feature_names] < train_min) | (synthetic_df[feature_names] > train_max)).sum()
display(pd.DataFrame({
    "Synthetic values outside training range": outside_range,
    "Total synthetic rows": len(synthetic_df),
}))

minority_class = y_train.value_counts().idxmin()
real_minority_scaled = X_train_scaled[np.asarray(y_train == minority_class)]
synthetic_minority_scaled = X_synthetic_scaled[np.asarray(y_synthetic == minority_class)]

if len(synthetic_minority_scaled) > 0:
    nn = NearestNeighbors(n_neighbors=1).fit(real_minority_scaled)
    distances, _ = nn.kneighbors(synthetic_minority_scaled)
    print(f"Rarest training class: {minority_class}")
    print("Median nearest-neighbour distance:", round(float(np.median(distances)), 4))
    print("95th percentile nearest-neighbour distance:", round(float(np.percentile(distances, 95)), 4))
else:
    print("No synthetic rows were generated for the rarest class.")

# %% [markdown]
# ## Output 7 — Class Distribution Comparison Graph
# **(Original vs Random Oversampling vs SMOTE)**

# %%
# ---------------------------------------------------------------------
# 7. Class distribution comparison: Original vs ROS vs SMOTE  [FIX #1]
# ---------------------------------------------------------------------
original_counts = y_train.value_counts().sort_index()
ros_counts      = pd.Series(y_ros).value_counts().sort_index()
smote_counts    = pd.Series(y_smote).value_counts().sort_index()

comparison_df = pd.DataFrame({
    "Original":            original_counts.values,
    "Random Oversampling": ros_counts.values,
    "SMOTE":               smote_counts.values,
}, index=original_counts.index)

ax = comparison_df.plot(kind="bar", figsize=(12, 6), width=0.8)
ax.set_title("Output 7: Class Distribution — Original vs Random Oversampling vs SMOTE")
ax.set_xlabel("Bean class")
ax.set_ylabel("Number of samples")
ax.legend(title="Strategy")
plt.xticks(rotation=35)
plt.tight_layout()
plt.savefig("class_distribution_comparison.png", dpi=300, bbox_inches="tight")
plt.show()

display(comparison_df)

# %% [markdown]
# ## Output 3–6 — Individual Model Results

# %%
# ---------------------------------------------------------------------
# 7b. Evaluation metrics  (with Minority PR-AUC added)  [FIX #4]
# ---------------------------------------------------------------------
print("\nMinority class used for minority precision/recall:", minority_class)

def calculate_metrics(model, X_eval, y_eval):
    y_pred = model.predict(X_eval)
    y_prob = model.predict_proba(X_eval)

    # Macro PR-AUC
    y_eval_binary = label_binarize(y_eval, classes=class_names)
    macro_pr_auc = average_precision_score(y_eval_binary, y_prob, average="macro")

    # Minority-class PR-AUC  [FIX #4 — added]
    # Use model.classes_ for correct column indexing  [FIX #2]
    minority_col_idx = list(model.classes_).index(minority_class)
    y_eval_minority_binary = (np.asarray(y_eval) == minority_class).astype(int)
    minority_pr_auc = average_precision_score(
        y_eval_minority_binary, y_prob[:, minority_col_idx]
    )

    # Per-class FPR (one-vs-rest)
    cm = confusion_matrix(y_eval, y_pred, labels=class_names)
    fprs = []
    for i in range(len(class_names)):
        tp = cm[i, i]
        fn = cm[i, :].sum() - tp
        fp = cm[:, i].sum() - tp
        tn = cm.sum() - (tp + fn + fp)
        fprs.append(fp / (fp + tn) if (fp + tn) > 0 else 0.0)

    return {
        "Accuracy": accuracy_score(y_eval, y_pred),
        "Macro Precision": precision_score(y_eval, y_pred, average="macro", zero_division=0),
        "Macro Recall": recall_score(y_eval, y_pred, average="macro", zero_division=0),
        "Macro-F1": f1_score(y_eval, y_pred, average="macro", zero_division=0),
        "Minority Precision": precision_score(y_eval, y_pred, labels=[minority_class], average="macro", zero_division=0),
        "Minority Recall": recall_score(y_eval, y_pred, labels=[minority_class], average="macro", zero_division=0),
        "Macro PR-AUC": macro_pr_auc,
        "Minority PR-AUC": minority_pr_auc,
        "Macro FPR": float(np.mean(fprs)),
    }

def evaluate_all(X_eval, y_eval):
    rows = []
    for name, model in models.items():
        row = {"Experiment": name}
        row.update(calculate_metrics(model, X_eval, y_eval))
        rows.append(row)
    return pd.DataFrame(rows).set_index("Experiment")

# %%
# ---------------------------------------------------------------------
# 8. Individual model results on validation set  [FIX #5 — Outputs 3-6]
# ---------------------------------------------------------------------
print("=" * 70)
print("OUTPUT 3: BASELINE MODEL RESULTS (Validation Set)")
print("=" * 70)
print(classification_report(y_val, baseline_model.predict(X_val_scaled),
                            labels=class_names, zero_division=0))

print("=" * 70)
print("OUTPUT 4: RANDOM OVERSAMPLING RESULTS (Validation Set)")
print("=" * 70)
print(classification_report(y_val, ros_model.predict(X_val_scaled),
                            labels=class_names, zero_division=0))

print("=" * 70)
print("OUTPUT 5: SMOTE RESULTS (Validation Set)")
print("=" * 70)
print(classification_report(y_val, smote_model.predict(X_val_scaled),
                            labels=class_names, zero_division=0))

print("=" * 70)
print("OUTPUT 6: CLASS WEIGHTING RESULTS (Validation Set)")
print("=" * 70)
print(classification_report(y_val, weighted_model.predict(X_val_scaled),
                            labels=class_names, zero_division=0))

# %%
# ---------------------------------------------------------------------
# 9. Combined validation results
# ---------------------------------------------------------------------
print("\nVALIDATION RESULTS")
validation_df = evaluate_all(X_val_scaled, y_val)
display(validation_df.round(4))

# %% [markdown]
# ## Output 9 — Model Comparison Table (Test Set)

# %%
# ---------------------------------------------------------------------
# 10. Final test results
# ---------------------------------------------------------------------
print("\nTEST RESULTS")
test_results_df = evaluate_all(X_test_scaled, y_test)
display(test_results_df.round(4))
test_results_df.to_csv("test_metrics_comparison.csv")
print("Saved: test_metrics_comparison.csv")

# %%
# ---------------------------------------------------------------------
# 11. Metric comparison plot
# ---------------------------------------------------------------------
metrics_to_plot = ["Macro-F1", "Minority Precision", "Minority Recall",
                   "Macro PR-AUC", "Minority PR-AUC"]
ax = test_results_df[metrics_to_plot].plot(kind="bar", figsize=(12, 6), ylim=(0, 1), rot=15)
ax.set_title("Test-set comparison of imbalance-handling strategies")
ax.set_ylabel("Score"); ax.set_xlabel("Training strategy")
ax.legend(loc="lower right")
plt.tight_layout()
plt.savefig("metric_comparison.png", dpi=300, bbox_inches="tight")
plt.show()

# %% [markdown]
# ## Output 8 — Confusion Matrices

# %%
# ---------------------------------------------------------------------
# 12. Confusion matrices
# ---------------------------------------------------------------------
fig, axes = plt.subplots(2, 2, figsize=(16, 13))
for ax, (name, model) in zip(axes.ravel(), models.items()):
    cm = confusion_matrix(y_test, model.predict(X_test_scaled), labels=class_names)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_names, yticklabels=class_names, ax=ax)
    ax.set_title(name)
    ax.set_xlabel("Predicted class"); ax.set_ylabel("Actual class")
    ax.tick_params(axis="x", rotation=40); ax.tick_params(axis="y", rotation=0)
plt.suptitle("Output 8: Confusion Matrices", fontsize=14, y=1.01)
plt.tight_layout()
plt.savefig("confusion_matrices.png", dpi=300, bbox_inches="tight")
plt.show()

# %%
# ---------------------------------------------------------------------
# 13. Per-class precision, recall and F1
# ---------------------------------------------------------------------
for name, model in models.items():
    print("\n" + "=" * 70 + f"\n{name}\n" + "=" * 70)
    print(classification_report(y_test, model.predict(X_test_scaled),
                                labels=class_names, zero_division=0))

# %% [markdown]
# ## Output 10 — Precision-Recall Curves

# %%
# ---------------------------------------------------------------------
# 14. Precision-recall curves for the minority class  [FIX #2]
# ---------------------------------------------------------------------
plt.figure(figsize=(8, 6))
y_test_minority_binary = (np.asarray(y_test) == minority_class).astype(int)

for name, model in models.items():
    # FIX #2: Use model.classes_ to find the correct column index per model
    minority_col_idx = list(model.classes_).index(minority_class)
    probs = model.predict_proba(X_test_scaled)[:, minority_col_idx]
    precision_vals, recall_vals, _ = precision_recall_curve(y_test_minority_binary, probs)
    ap = average_precision_score(y_test_minority_binary, probs)
    plt.plot(recall_vals, precision_vals, label=f"{name} (AP={ap:.3f})")

plt.title(f"Output 10: Precision-Recall Curves — Minority Class ({minority_class})")
plt.xlabel("Recall"); plt.ylabel("Precision")
plt.legend(); plt.grid(alpha=0.3); plt.tight_layout()
plt.savefig("minority_precision_recall.png", dpi=300, bbox_inches="tight")
plt.show()

# %%
# ---------------------------------------------------------------------
# 15. Training cost: inference latency and model size
# ---------------------------------------------------------------------
sample_size = min(500, len(X_test_scaled))
X_timing = X_test_scaled[:sample_size]
computational_rows = []

for name, model in models.items():
    model.predict_proba(X_timing[:min(20, sample_size)])  # warm-up
    repeats = 20
    start = time.perf_counter()
    for _ in range(repeats):
        model.predict_proba(X_timing)
    elapsed = time.perf_counter() - start
    computational_rows.append({
        "Experiment": name,
        "Inference latency (ms/sample)": (elapsed / repeats / sample_size) * 1000,
        "Model size (MB)": len(pickle.dumps(model)) / (1024 ** 2),
    })

computational_df = pd.DataFrame(computational_rows).set_index("Experiment")
display(computational_df.round(4))
computational_df.to_csv("computational_cost_comparison.csv")
print("Saved: computational_cost_comparison.csv")

# %% [markdown]
# ## Output 11 — Error Analysis

# %%
# ---------------------------------------------------------------------
# 16. Representative test errors
# ---------------------------------------------------------------------
prediction_examples = X_test_df.copy().reset_index(drop=True)
prediction_examples["Actual class"] = np.asarray(y_test)
for name, model in models.items():
    prediction_examples[name + " prediction"] = model.predict(X_test_scaled)

incorrect_mask = np.zeros(len(prediction_examples), dtype=bool)
for name in models:
    incorrect_mask |= (prediction_examples[name + " prediction"].to_numpy()
                       != prediction_examples["Actual class"].to_numpy())

error_examples = prediction_examples.loc[incorrect_mask].head(10)
display(error_examples)
error_examples.to_csv("representative_test_errors.csv", index=False)
print("Saved: representative_test_errors.csv")

# %%
# ---------------------------------------------------------------------
# 17. Reproducibility information
# ---------------------------------------------------------------------
print("\nPython version:", sys.version)
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Scikit-learn version:", sklearn.__version__)
print("Imbalanced-learn version:", imblearn.__version__)
print("Random seed:", RANDOM_STATE)
print("Train/validation/test split: 70% / 15% / 15%")
print("Classifier: RandomForestClassifier (300 trees)")
print("SMOTE k_neighbors:", smote_k)
print("Minority class:", minority_class)
print("\nOriginal class distribution:")
display(y.value_counts().sort_index().to_frame("Count"))
```
